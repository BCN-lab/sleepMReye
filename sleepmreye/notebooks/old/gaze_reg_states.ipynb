```python
from mreyemove.analysis.glm.config import GLMConfig
from SleepMReye.sleepmreye.sleep_mrio import SleepMRIO
from mreyemove.analysis.glm.cli import construct_desc

mrio = SleepMRIO(
    bids_root="/Users/zach/2025_RA/matthias/bids_sleepmreye",
    exclude_subjects=["01", "07", "09", "12"],
    tr=2.1,
    task="sleep",
    load_eog=False,
    desc_add="clean-with-fd-clamped"
)

config = GLMConfig.from_yaml("/Users/zach/2025_RA/matthias/eyemove_and_sleep/SleepMReye/sleepmreye/designs/original_sleep.yml")
desc_add = construct_desc(dataset=mrio, config=config)
```

```python
import pandas as pd

subject_gaze_bins = []

for subject in mrio.subjects:
    bins_sleep = []
    bins_wake = []
    for _, run, run_df in mrio.iter_run_events(subject=subject):
        # Extract gaze reg 
        gaze_reg_sleep = run_df["mreyemove_S"]
        gaze_reg_wake = run_df["mreyemove_W"]
        
        # Bin the data 
        bins_sleep.append(pd.qcut(gaze_reg_sleep, q=2, labels=['low', 'high']))
        bins_wake.append(pd.qcut(gaze_reg_wake, q=2, labels=['low', 'high']))
        
    subject_gaze_bins.append({
        "subject": subject,
        "n_runs": max(len(bins_wake), len(bins_sleep)),
        bins_sleep
    })
    
```
